# Data Cleaning

```python
# Load the files again
import pandas as pd
import numpy as np

sales_df = pd.read_csv("retail_sales_ml_apl.csv")
inventory_df = pd.read_csv("retail_inventory_ml_apl.csv")
```

```python
# Make copies before cleaning
sales_clean = sales_df.copy()
inventory_clean = inventory_df.copy()
```

```python
# Clean column names and text values to remove extra spaces and makes the text more consistent.

# Remove extra spaces from column names
sales_clean.columns = sales_clean.columns.str.strip()
inventory_clean.columns = inventory_clean.columns.str.strip()

# Clean text columns
for df in [sales_clean, inventory_clean]:
    text_cols = df.select_dtypes(include="object").columns
    for col in text_cols:
        df[col] = df[col].astype("string").str.strip()
```

```python
# Convert date columns

# Sales date
sales_clean["Transaction Date"] = pd.to_datetime(sales_clean["Transaction Date"], errors="coerce")

# Inventory dates
inventory_clean["Start Date"] = pd.to_datetime(inventory_clean["Start Date"], errors="coerce")

# Replace 9999-12-31 with NaT because it usually means ongoing/current
inventory_clean["End Date"] = inventory_clean["End Date"].replace("9999-12-31", pd.NaT)
inventory_clean["End Date"] = pd.to_datetime(inventory_clean["End Date"], errors="coerce")
```

```python
# Handle the empty Reason of Return column since this column is completely empty, the cleanest option is to remove it.
if sales_clean["Reason of Return"].isna().all():
    sales_clean = sales_clean.drop(columns=["Reason of Return"])
```

```python
# Ensure numeric columns are really numeric

# Sales numeric columns
sales_numeric_cols = ["Is Return", "Qty Sold", "Sales Amount", "Cogs", "Number of Transactions"]
for col in sales_numeric_cols:
    sales_clean[col] = pd.to_numeric(sales_clean[col], errors="coerce")

# Inventory numeric columns
inventory_numeric_cols = [
    "Qty on hand",
    "Stocks Selling Amount",
    "Cost of Stocks",
    "Stock Unit Selling Price",
    "Stock Unit Cost Price"
]
for col in inventory_numeric_cols:
    inventory_clean[col] = pd.to_numeric(inventory_clean[col], errors="coerce")
```

```python
# Remove duplicate rows even though my files already have none, this is still a good cleaning step.
sales_clean = sales_clean.drop_duplicates()
inventory_clean = inventory_clean.drop_duplicates()
```

```python
# Check the cleaned data again
print("Sales shape:", sales_clean.shape)
print("Inventory shape:", inventory_clean.shape)

print("\nSales missing values:")
print(sales_clean.isna().sum()[sales_clean.isna().sum() > 0])

print("\nInventory missing values:")
print(inventory_clean.isna().sum()[inventory_clean.isna().sum() > 0])

print("\nSales data types:")
print(sales_clean.dtypes)

print("\nInventory data types:")
print(inventory_clean.dtypes)
```

# Data Exploration and Visualization

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
sales = sales_clean.copy()
inventory = inventory_clean.copy()
```

```python
# Create a monthly sales summary, this helps see trends over time.
sales["Month"] = sales["Transaction Date"].dt.to_period("M").dt.to_timestamp()

monthly_sales = sales.groupby("Month").agg(
    Sales_Amount=("Sales Amount", "sum"),
    Qty_Sold=("Qty Sold", "sum"),
    Returns=("Is Return", "sum"),
    Transactions=("Number of Transactions", "sum")
).reset_index()

monthly_sales
```

```python
# Line chart of sales over time, this reveals whether sales are increasing, decreasing, or fluctuating month by month.
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales["Month"], monthly_sales["Sales_Amount"], marker="o")
plt.title("Monthly Sales Amount")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Bar chart of sales by Sales Type, this tells which sales type contributes the most revenue.
sales_type_summary = sales.groupby("Sales Type")["Sales Amount"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(sales_type_summary.index, sales_type_summary.values)
plt.title("Sales Amount by Sales Type")
plt.xlabel("Sales Type")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
# Bar chart of top product categories by sales, this helps identify which product categories are performing best.
top_categories = sales.groupby("Product Category")["Sales Amount"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(top_categories.index, top_categories.values)
plt.title("Top 10 Product Categories by Sales Amount")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.xticks(rotation=75)
plt.show()
```

```python
# Bar chart of top stores by sales, this helps compare store performance.
top_stores = sales.groupby("Store")["Sales Amount"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(top_stores.index, top_stores.values)
plt.title("Top 10 Stores by Sales Amount")
plt.xlabel("Store")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
# Pie chart of return vs non-return transactions, this gives a quick view of how common returns are.
return_counts = sales["Is Return"].value_counts().sort_index()

plt.figure(figsize=(6, 6))
plt.pie(return_counts, labels=["Not Return", "Return"], autopct="%1.1f%%", startangle=90)
plt.title("Return vs Non-Return Transactions")
plt.show()
```

```python
# Inventory stock status chart, This helps see how inventory is distributed across status levels like Full Price, Markdown, Clearance, and Promo.
stock_status_summary = inventory.groupby("Stock Status")["Qty on hand"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(stock_status_summary.index, stock_status_summary.values)
plt.title("Inventory Quantity by Stock Status")
plt.xlabel("Stock Status")
plt.ylabel("Qty on hand")
plt.xticks(rotation=45)
plt.show()
```

```python
# Inventory quantity by product category, this helps see which categories have the most stock available.
inventory_category_summary = inventory.groupby("Product Category")["Qty on hand"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(inventory_category_summary.index, inventory_category_summary.values)
plt.title("Top 10 Product Categories by Quantity on Hand")
plt.xlabel("Product Category")
plt.ylabel("Qty on hand")
plt.xticks(rotation=75)
plt.show()
```

```python
# Inventory selling amount by stock status, This helps compare the value of stock across different stock statuses.
stock_amount_summary = inventory.groupby("Stock Status")["Stocks Selling Amount"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(stock_amount_summary.index, stock_amount_summary.values)
plt.title("Stocks Selling Amount by Stock Status")
plt.xlabel("Stock Status")
plt.ylabel("Stocks Selling Amount")
plt.xticks(rotation=45)
plt.show()
```

# Conclusions
The sales dataset shows a healthy overall performance:

- Total Sales Amount is about 10.48 million
- Total COGS is about 5.85 million
- Estimated gross margin is about 4.63 million, or roughly 44.2%
- Return rows make up about 5.96% of transactions, so returns exist but are not overwhelming

The strongest sales pattern is that Full Price dominates sales. It contributes the biggest share of revenue and transactions, much more than Promo, Markdown, or Clearance items. This suggests that most revenue is coming from regular-priced products rather than discounted ones.

The best-performing product groups are mostly in footwear, especially Scholar Footwear and Femme Footwear. That means the business seems strongest in those lines, while other divisions like Junior Apparel contribute much less.

There is also a time trend in sales. Your sales appear to peak around December 2025, while some months like January 2026 are lower. That suggests seasonality or holiday-driven demand.

In the inventory file, most stock is also under Full Price status. This matches the sales pattern and suggests the business is carrying a large amount of full-price inventory, which is good if those items keep selling well. The inventory also has a lot of rows with an open-ended End Date, which likely means those items are still active.
